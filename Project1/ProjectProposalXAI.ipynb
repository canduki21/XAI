## 1. Goals of the Study

I will fine-tune a pretrained ResNet-18 convolutional neural network to classify lunar regolith simulants (e.g., LHS-1, LMS-1) using surface texture imagery.

### Explanation Problem
Lunar simulants are visually similar gray granular materials. A model may rely on shortcut features such as lighting, background artifacts, or the presence of a scale cube instead of true geological markers.

### Motivation
For autonomous excavation and construction systems under MATEAR, incorrect material identification could lead to structural failure or inefficient ISRU processing. Therefore, I aim to ensure the model bases its decisions on meaningful geological features such as:

- Grain size distribution  
- Albedo variations  
- Particle angularity  

rather than spurious photographic artifacts.



## 2. Data and Model Plans

### Dataset
High-resolution imagery of lunar simulants from Space Resource Technologies.

### Preprocessing
- Image tiling into 224 × 224 patches  
- Brightness and contrast augmentation  
- Normalization  
- Random horizontal flips  

### Model
Pretrained ResNet-18 fine-tuned for multi-class simulant classification.


```python

import torch
import torch.nn as nn
import torchvision.models as models

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

model = models.resnet18(pretrained=True)
model.fc = nn.Linear(model.fc.in_features, 8)  # Adjust class count as needed
model = model.to(device)

print("Model ready for training.")
```


## 3. Explainability Methods
